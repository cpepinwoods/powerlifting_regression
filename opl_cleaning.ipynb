```python
import pandas as pd
import numpy as np
from tqdm import tqdm
```

```python
# read in the data
df = pd.read_csv('data/openpowerlifting.csv')
df.head()
```

```python
df = df[df["Event"] == "SBD"]
df = df[df["Tested"] == "Yes"]
# TODO set this to be only raw/single/wraps
df.shape
```

```python
df.drop(["Event", "Squat4Kg", "Bench4Kg", "Deadlift4Kg", "Tested", 
         "MeetName", "Wilks", "Glossbrenner", "Goodlift", "Sanctioned",
         "AgeClass", "BirthYearClass", "Country", "State", "MeetCountry",
         "MeetState", "MeetTown"], axis=1, inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
df = df.drop_duplicates(subset=["Name", "Date", "Federation"])
df.shape
```

```python
df = df[df["TotalKg"] > 0]
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Sex'] = le.fit_transform(df['Sex'])
df['Equipment'] = le.fit_transform(df['Equipment'])
df['Division'] = le.fit_transform(df['Division'])
df['Federation'] = le.fit_transform(df['Federation'])
df['ParentFederation'] = le.fit_transform(df['ParentFederation'])
df['Place'] = le.fit_transform(df['Place'])
df['WeightClassKg'] = le.fit_transform(df['WeightClassKg'])
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```
