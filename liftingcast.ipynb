```python
def liftingcast_fromfile(filename):
    lifters = dict()
    with open("data/" + filename) as file:
        soup = BeautifulSoup(file, "html.parser")
        # Find the table with the results
        children = soup.find_all("a")
        for child in children:
            link = child.get('href')
            
            # If the link is for a division then parse it and set it as a key
            if "division" in link:
                div = str(child).split(">")
                division = div[1][:-4]
                wc = div[2][:-3]
                div = division + " " + wc
                lifters[div] = []

            # If the link is for a lifter then add their name as a value in the dictionary for the current key
            if "lifter" in link:
                lifters[div].append(child.get_text())
        
    return lifters

def liftingcast_fromlink(link):
    import requests
    response = requests.get(link)
    html = response.text

    lifters = dict()
    soup = BeautifulSoup(html, "html.parser")
    # Find the table with the results
    children = soup.find_all("a")
    for child in children:
        link = child.get('href')
        
        # If the link is for a division then parse it and set it as a key
        if "division" in link:
            div = str(child).split(">")
            division = div[1][:-4]
            wc = div[2][:-3]
            div = division + " " + wc
            lifters[div] = []

        # If the link is for a lifter then add their name as a value in the dictionary for the current key
        if "lifter" in link:
            lifters[div].append(child.get_text())
    
    return lifters
```

```python
opl = None
def load_open_pl():
    global opl
    opl = pd.read_csv("data/openpowerlifting.csv")
```

```python
def openpl_lookup(name):
    return opl[opl["Name"] == name]
```

```python
def get_div_avg(division):
    return 1

def mlp_format(results):
    return (1, 1)

def run_mlp(results, n):
    return np.random.randint(100)
```

```python
def get_predicted_total(results, division):
    if len(results) == 0:
        return get_div_avg(division)
    else:
        results, n = mlp_format(results)
        return run_mlp(results, n)
```

```python
for division in lifters:
    print(division)
    lr = []
    for lifter in lifters[division]:
        results = openpl_lookup(lifter)
        pred_total = get_predicted_total(results, division)
        lr.append((lifter, pred_total))
    lr.sort(key=lambda x: x[1])
    lifters[division] = lr

print(lifters)        
```
